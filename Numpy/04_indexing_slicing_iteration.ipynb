# Iteration
Iteration means visiting each element of a NumPy array one by one using a loop.

# Indexing
Indexing means accessing a specific element from a NumPy array using its position (index).


# Slicing
Slicing means extracting a part/range of elements from a NumPy array.

# Reshaping an Numpy array

```python
import numpy as np
```

```python
array_a = np.arange(24)
array_a
```

```python
array_a.reshape(6,4)
```

# Indexing of an array OR extracting elements using the index 

# For 1-D array

```python
array_a
```

```python
arr = array_a.copy()
arr
```

```python
print(arr[14])
arr[14]
```

```python
print(arr[2:6])
arr[2:6]
```

```python
print(arr[2:10:2])
arr[2:10:2]
```

# For 2-D array

```python
arr1 = np.array([[1,2,3],[2,3,4],[3,4,5]])
arr1
```

```python
print(arr1[2])
arr1[2]
```

```python
# suppose we only extract the 2nd element of every array from the arr1
print(arr1[:,1])
arr1[:,1]
```

```python
print(arr1[:,0:2])
arr1[:,1:3]
```

# Iteration


```python
for i in arr1:
    print (i)
```

```python
for i in np.nditer(arr1):
    print(i)
```
