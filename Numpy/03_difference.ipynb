# Numpy Array VS Python List




```python
import numpy as np
```

# concatination

```python
#  list
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)
```

```python
#  numpy array
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)
```

# multiplication

```python
# list 
a = [1, 2, 3]
print(a * 2)
```

```python
# numpy array
a = np.array([1, 2, 3])
print(a * 2)
```

# 2D Data

```python
# list
matrix = [[1, 2], [3, 4]]
matrix
```

```python
# numpy array
matrix = np.array([[1, 2], [3, 4]])

print(matrix.shape)
```

```python
import time
```

```python
# for list 
x = range(10000000)
y = range(10000000,20000000)

start_time = time.time()

c = [(x,y) for x,y in zip(x,y)]
print(time.time() - start_time)
```

```python
# for numpy
a = np.arange(10000000)
b = np.arange(10000000,20000000)

start_time = time.time()
c = a+b

print(time.time() - start_time)
```
