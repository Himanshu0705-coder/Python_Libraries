# Reshaping Numpy Array

1. Ravel    -- it basically convert n-d arrays into 1-d arrayx
2. Reshape
3. Transpose
4. Stacking
5. Splitting

```python
import numpy as np
```

```python
arr1= np.array[[1,2,3],[3,4,5],[4,5,6]]
arr1
```

```python
arr1.shape
```

# Ravel

```python
arr1.ravel()
```

# Transpose

```python
arr1.transpose()
```

# Stacking

```python
arr2 = np.ones((3,3))
arr2
```

```python
# hstack -- horizontal stack
np.hstack((arr1,arr2))
```

```python
# vstack -- vertical stack
np.vstack((arr1,arr2))
```

# Splitting

```python
# hsplit  -- horizontal split
np.hsplit(arr1,3)
```

```python
#  vsplit  -- vertical split
np.vsplit(arr1,3)
```
