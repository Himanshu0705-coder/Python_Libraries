# Broadcasting

The term broadcasting refers to the ability of Numpy od treat arrays of different shapes during arithmetic operations. Arithmetic operations on array are usually done on corresponding elements.If two arrays are of exactly the same shape,these operations are smoothly performed.

If the dimensions of two arrays are disimilar,element-to-elemnt operations are not possible. However, operations on arrays of non-similar shapes is still possible in Numpy, because of the broadcasting capability. The smaller arrays is broadcast to the size of the larger array , so that they have compatible shapes.




```python
import numpy as np
```

```python
# Scenerio 1
a1 = np.arange(8).reshape(2,4)
a2 = np.arange(8,16).reshape(2,4)

print(a1)
print(a2)
```

```python
a1 + a2
```

```python
# Scenerio 2
arr3 = np.arange(9).reshape(3,3)
arr4 = np.arange(3).reshape(1,3)

print(arr3,arr4)
```

```python
arr3 + arr4
```

```python
# If x=1 and y=n then also operation will take place(Same Dimension)
a5 = np.arange(3).reshape(1,3)
a6 = np.arange(12).reshape(4,3)

print(a5)
print(a6)
```

```python
a5 + a6
```

```python
a9 = np.arange(3).reshape(1,3)
a10 = np.arange(16).reshape(4,4)


print(a9)
print(a10)
```

```python
a9 + a10
```

```python
a11 = np.arange(3).reshape(1,3)
a12 = np.arange(3).reshape(3,1)

print(a11)
print(a12)
```

```python
a11 + a12
```

```python
# If x=1 and y=1,then the operation will take place no matter what

a13 = np.arange(1).reshape(1,1)
a14 = np.arange(20).reshape(4,5)

print(a13)
print(a14)
```

```python
a13 + a14
```

```python
# If they are of different dimensions

a15 = np.arange(4)
a16 = np.arange(20).reshape(5,4)

print(a15)
print(a16)
```

```python
a15 + a16
```
