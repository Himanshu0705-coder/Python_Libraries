# Flipkart Scaping data using Beautifulsoup

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
web = requests.get("https://www.amazon.in/s?k=phone&crid=20ZFSXPR0MJ64&sprefix=pho%2Caps%2C327&ref=nb_sb_noss_2")
web
```

```python
soup = BeautifulSoup(web.text,"html.parser")
soup
```
