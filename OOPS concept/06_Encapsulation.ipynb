# Encapsulation  
- Encapsulation means combining data (variables) and methods (functions) inside a class and controlling how that data is accessed or modified.

- we use encapsulation to protect data using `private attribute` , and after that set the getter and setter methods there to aceess and update the data 

# Encapsulation concept
- Need for encapsulation
- Private attribute
- getter and setter methods
- class diagram

```python
class ATM:

    # special/magic/dunder methods
    #Costructor

    # instance variable
    
    def __init__(self):
        self.__pin = ""
        self.__balance = 0

        self.__menu()

    # getter and setter
    def get_pin(self):
        return self.__pin

    def set_pin(self,new_pin):
        if type(new_pin) == str:
            self.__pin = new_pin
            print("Pin changed")
        else:
            print("Not allowed")
        
    def __menu(self):
        user_input = input("""
                        Hello, how would you like to proceed?
                        1, Enter 1 to create pin
                        2. Enter 2 to deposit
                        3. Enter 3 to withdraw
                        4. Enter 4 to check balance
                        5. Enter 5 to exit
        """)

        if user_input == "1":
            self.create_pin()

        elif user_input == "2":
            self.deposit()

        elif user_input =="3":
            self.withdraw()

        elif user_input =="4":
            self.check_balance()

        else:
            print("bye")

    def create_pin(self):
        self.__pin = input("Enter your Pin")
        print("Pin is successful")


    def deposit(self):
        temp = input("Enter the value")
        if temp == self.__pin:
            amount = int(input("Enter the amount"))
            self.__balance = self.__balance + amount
            print("Deposit Successful")
        else:
            print("invalid pin")


    def withdraw(self):
        temp = input("Enter the value")
        if temp == self.__pin:
            amount = int(input("Enter the amount"))
            if amount <= self.__balance:
                self.__balance = self.__balance - amount
                print("Operation successful")
            else:
                print("Insufficient funds")
        else:
            print("invalid pin")


    def check_balance(self):
        temp = input("Enter the value")
        if temp == self.__pin:
            print(self.__balance)
        else:
            print("invalid pin")

    
```
