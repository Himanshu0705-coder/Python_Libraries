```python
class Customer:
    def __init__(self,name):
        self.name = name
    
    def greet(customer):
        print("Hello",customer.name)

    


```

```python
cust = Customer("Himanshu")
cust
```

```python
cust.greet()
```

```python
class Customer:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender

    def greet(customer):
        if customer.gender == "Male":
            print("Hello!",customer.name,"sir")
        else:
            print("Hello!",customer.name,"ma'am")
        
```

```python

cust = Customer("Aastha","Female")
cust1 = Customer("Himanshu","Male")
```

```python
cust.greet()
cust1.greet()
```
