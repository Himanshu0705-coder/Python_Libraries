# Collection of Objects

```python
class Customer:
    def __init__(self,name,age):
        self.name= name
        self.age = age
    
```

```python
c1 = Customer("Himanshu",21)
c2 = Customer("Aastha",19)
c3 = Customer("Aaditya",25)
```

```python
L = [c1,c2,c3]
```

```python
for i in L:
    print(i.name,i.age)
    
```

```python
class Names:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def intro(self):
        print("I'm",self.name,"and I'm",self.age,"years old")
        
```

```python
c1 = Names("Himanshu",21)
c2 = Names("Aastha",19)
c3 = Names("Aaditya",25)
```

```python
L = [c1,c2,c3]


for i in L:
    i.intro()
```
