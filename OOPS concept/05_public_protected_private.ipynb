- Python OOP (Object-Oriented Programming) mein `Public`, `Protected` and `Private` ka use class ke variables aur methods ki accessibility ko indicate karne ke liye hota hai.

# Public variable
- Public variables ko class ke bahar bhi directly access kar sakte hain

```python
class Student:
    def __init__(self):
        self.name = "Himanshu"

s = Student()

print(s.name)
```

# Protected variable
- Protected variable ke naam ke aage single underscore _ hota hai.

```python
class Student:
    def __init__(self):
        self._marks = 90

s = Student()

print(s._marks)
```

# Private variable
Private variable ke naam ke aage double underscore __ hota hai.

```python
class Student:
    def __init__(self):
        self.__marks = 90

s = Student()

# print(s.__marks)  # AttributeError
print(s._Student__marks)
```
