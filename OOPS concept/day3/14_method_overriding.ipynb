# Method Overriding

- This concept comes inside the polymorphism.

Method overriding occurs when a `child class` defines a method with the same name as a method in its `parent class`, providing its own implementation.

`Parent class` ka method `child class` mein same name se dobara define karna, taaki child apna behaviour provide kar sake.

```python
class Parent:
    def salary(self):
        print("Parents salary")

class child(Parent):
    def salary(Self):
        print("children salary")
```

```python
amount.salary()
```
