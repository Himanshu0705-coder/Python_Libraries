# method Overloading


- it is also the concept of Polymorphism.

technicALLLY `python` doesn't support method overloading

```python
class Geometry:
    def area(self,radius):
        print("circle",3.14*radius*radius)

    def area(self,a,b):
        print("Rectangle",a*b)

obj = Geometry()
obj.area(3)



# it will through error because it override the first `area()` function
```

```python
class Geometry:
    def area(self,a,b=0):
        if b ==0:
            print("Circle",3.14*a*a)
        
        else:
            print("Rectangle", a *b)


obj = Geometry()

obj.area(3)
obj.area(3,6)
```
