# Method Resolution Order (MRO)

If both parent classes contain a method with the same name, Python uses MRO to determine which implementation to call.

```python
class A:
    def show(self):
        print("Class A")


class B:
    def show(self):
        print("Class B")


class C(A, B):
    pass


obj = C()
obj.show()
```

```python
print(C.mro())
```
