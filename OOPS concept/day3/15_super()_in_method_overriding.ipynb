# Super function()

it call the parent function and child function one by one .

- firstly, it call the child class function . Then call parents class function.

```python
class Phone:
    def __init__(self,price,brand,camera):
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print('Buying phone')


class SmartPhone(Phone):
        
    def buy(self):
        print("Buying a smartphone")
        super().buy()
             
s=SmartPhone(20000,"Apple", 13)
```

```python
s.buy()
```

```python
class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):

    def __init__(self,price,brand,camera,os,ram):
        print("Phle ye ")

        super().__init__(price,brand,camera)
        self.os = os
        self.ram = ram
        print("Inside smartphone constructor")


s = SmartPhone(20000,"Smartphone",12,"Android",2)

print(s.os)
print(s.brand)

```
