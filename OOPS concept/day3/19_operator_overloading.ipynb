# Operator Overloading


Operator overloading means giving an operator different behaviour depending on the objects or data types on which it operates.

- For example, the + operator adds numbers but joins strings.

```python
print(10 + 20)           # Addition
print("Hello " + "World") # String concatenation
```

# Operator overloading using a class

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```
