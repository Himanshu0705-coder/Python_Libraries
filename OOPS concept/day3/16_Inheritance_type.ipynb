# Inheritance types

- single level inheritance
- multi level inheritance
- multiple level inheritance
- hierarchical inheritance

# single inheritance 


```python
class Animal:
    def eat(self):
        print("Animal can eat")


class Dog(Animal):
    def bark(self):
        print("Dog can bark")


d = Dog()
d.eat()
d.bark()
```

# Multilevel inheritance


```python
class Grandparent:
    def house(self):
        print("Grandparent owns a house")


class Parent(Grandparent):
    def car(self):
        print("Parent owns a car")


class Child(Parent):
    def bike(self):
        print("Child owns a bike")


obj = Child()

obj.house()
obj.car()
obj.bike()
```

# Hierarchical inheritance



```python
class Animal:
    def eat(self):
        print("Animals can eat")


class Dog(Animal):
    def bark(self):
        print("Dog can bark")


class Cat(Animal):
    def meow(self):
        print("Cat can meow")


class Cow(Animal):
    def moo(self):
        print("Cow can moo")


d = Dog()
d.eat()
d.bark()

c = Cat()
c.eat()
c.meow()

co = Cow()
co.eat()
co.moo()
```

# Multiple inheritance


```python
class A:
    def show(self):
        print("Class A")


class B:
    def show(self):
        print("Class B")


class C(A, B):
    pass


obj = C()
obj.show()
```
