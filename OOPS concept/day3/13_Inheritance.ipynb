# Inheritance

`Inheritance` is an OOP concept in which one class acquires the properties and methods of another class.

- inheritance provide the `code reusability` property.

```python
class User:
    def login(self):
        print("login")

    def register(self):
        print("register")

    def student(sef):
        print("Enroll")

    def review(self):
        print("Review")
```

```python
std1 = User()     # object of class


std1.login()
std1.register()
std1.review()
std1.student()
```
