# static/Class variable

A `static variable` is a variable declared inside a class but outside its methods. In Python, it is commonly called a `class variable.`

- it means static variable class ke andr create krte h .. but constructor se bahar.

- static variable are same for all the users.

```python
class Student:
    # static variable/class variable
    college = "Shobhit University"   

    #  instance variable 
    def __init__(self,name):
        self.name = name  # instance variable
    

    

```

```python
s1 = Student("Himanshu")
print("My name is ",s1.name,"and I'm studying in ",s1.college)
```

```python
class Student:
    # static variable/class variable
    __college = "Shobhit University"   

    #  instance variable 
    def __init__(self,name):
        self.name = name  # instance variable
    
    
    @staticmethod
    def get_college():
        return Student.__college
```

```python
s1 = Student("Himanshu")
print("My name is ",s1.name,"and I'm studying in ",Student.get_college())
```
