# sort values()

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('Datasets/matches.csv')
df.head(1)
```

```python
(df['team1'].value_counts() + df['team2'].value_counts()).sort_values()
```

```python
data1 = df['team1'].value_counts()
data2 = df['team2'].value_counts()
```

```python
data1 + data2
```

```python
(data1 + data2).sort_values()
(data1 + data2).sort_values(ascending=False)
# (data1 + data2).sort_values().head()
# (data1 + data2).sort_values().tail()
```

```python
df.sort_values('city',ascending=False)
# df.sort_values('city',ascending=False).head(5)
```

# inplace 
it makes the permanent changes 


```python
df.sort_values('id', ascending=True ,inplace = True)
```
