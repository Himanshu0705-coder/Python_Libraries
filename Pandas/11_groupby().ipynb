```python
df.groupby('season')['win_by_runs'].sum().reset_index()
```

```python
#  add a new columns in the end
df["Avg_runs"] = df.groupby('season')['win_by_runs'].transform('mean')
df
```

# conditions

```python
df["Avg_runs"] = (
    df.groupby("season")["win_by_runs"]
      .transform("mean")
)
```

```python
df["Above_Avg"] = df["win_by_runs"] > df["Avg_runs"]
df["Above_Avg"]
```
