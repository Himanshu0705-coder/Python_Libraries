# Creat a function for extracting the Match cities

```python
def get_city_match_count(city):
    mask = (df['city']==city)
    return df[mask]
```

```python
get_city_match_count('Delhi').shape[0]
```

using two conditions 


```python
def get_city_date(city,date):
    mask1 = (df['city']==city)
    mask2 = (df['date']>date)
    return df[mask1 & mask2]
    # return df[mask1 | mask2]
```
