# Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(df['Category'],df['Quantity'])

plt.subplot(2,2,2)  
plt.plot(df['Category'],df['OrderValue'])

plt.subplot(2,2,3)
plt.plot(df['Quantity'],df['Category'])
plt.subplot(2,2,4)
plt.plot(df['PaymentMethod'],df['Status'])


```

# Object oriented way

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches spiis the dots-per-inch (pixel per inch).

```python
#  it is used to add make the axis size according to given labels
fig = plt.figure(figsize=(10,5) ,dpi=400)    # bydefault dpi(dot per inch) = 100     ---- it inhance the quality of graphs 
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.3,0.3,0.5])
axis2.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(12,6))

ax.plot(xaxis, xaxis+1, color="red", linewidth=0.25)
ax.plot(xaxis, xaxis+2, color="red", linewidth=0.50)
ax.plot(xaxis, xaxis+3, color="red", linewidth=1.00)
ax.plot(xaxis, xaxis+4, color="red", linewidth=2.00)

# possible linestyle options '-', '--', '-.', ':', 'steps'
ax.plot(xaxis, xaxis+5, color="green", lw=3, linestyle='-')
ax.plot(xaxis, xaxis+6, color="green", lw=3, ls='-.')
ax.plot(xaxis, xaxis+7, color="green", lw=3, ls=':')

# custom dash
line, = ax.plot(xaxis, xaxis+8, color="black", lw=1.50)
line.set_dashes([5, 15, 50, 15])  # format: line length, space length, ...

# possible marker symbols: marker='+', 'o', '*', 's', ',', '.', '1', '2', '3', '4', ...
ax.plot(xaxis, xaxis+9, color="blue", lw=3, ls='-', marker='+')
ax.plot(xaxis, xaxis+10, color="blue", lw=3, ls='-', marker='o')
ax.plot(xaxis, xaxis+11, color="blue", lw=3, ls='-', marker='s')
ax.plot(xaxis, xaxis+12, color="purple", lw=1, ls='-', marker='o', markersize=2)
ax.plot(xaxis, xaxis+13, color="purple", lw=1, ls='-', marker='o', markersize=4)
ax.plot(xaxis, xaxis+14, color="purple", lw=1, ls='-', marker='o', markersize=8,
        markerfacecolor="red")
ax.plot(xaxis, xaxis+15, color="purple", lw=1, ls='-', marker='o', markersize=8,
        markerfacecolor="yellow", markeredgewidth=3, markeredgecolor="green")
```
