# boxplot

```python
sns.boxplot(x=df['tip'],y=df['day'],data=df , palette='rainbow')
```

# Boxplot and Violinplot
boxplot and violinplot are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative in a way that facilitates comparisions between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be "outliers" using a method that is a function of the inter-quartile range.

```python
sns.violinplot(x=df['sex'],y=df['tip'])
```

```python
sns.violinplot(x=df['day'],y=df['total_bill'])
```

# strippot and swarmplot

# stripplot
stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box plot or violin plot in caases where you want to show all observations along with some represenation of the underlying distribution.

# swarmplot 
The swarmplot is similar to stripplot() , but the points are adjusted(only along the categorucal axis) so that they don't overlap. This gives a better representation of the distribution of values, although it doese not scale as well to large numbers of observations (both in terns of the ability to show all points and in tern of the computation needed to arrange them).

```python
sns.stripplot(x=df['day'],y=df['total_bill'],data=df)
```

```python
sns.swarmplot(x=df['day'],y=df['total_bill'],data=df)
```

```python
sns.violinplot(x=df['tip'],y=df['day'])
# sns.stripplot(x=df['tip'],y=df['day'])
sns.swarmplot(x=df['tip'],y=df['day'])
```
