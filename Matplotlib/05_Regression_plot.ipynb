# lmplot
regression plot sus used in machine learning so we will not fully discuss what, how and where it is used but we will see how it is created and what are the values we can put inside it.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
sns.lmplot(x='total_bill',y='tip',data = tips)
```

```python
sns.lmplot(x='total_bill',y='tip',hue ='sex',data = tips)
```

```python
# sns.lmplot(x='total_bill',y='tip',data = tips,hue ='sex',palette='bright')

# sns.lmplot(x='total_bill',y='tip',data = tips,hue ='sex',palette='bright',markers=['o','v'])

sns.lmplot(x='total_bill',y='tip',data = tips,hue ='sex',palette='bright',markers=['o','v'],
           scatter_kws={'s':50})
```
