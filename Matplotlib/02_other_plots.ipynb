We use distribution plots to visualize the distributiom of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a dataset. These plots are:

- displot/histplot
- jointplot
- pairplot
- rugplot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#  we can directly load the dataset usning below function
# seaborn also have its own datasets ..whom we can load directly
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

NOTE :- 
displot and histplot are same

```python
sns.displot(df['total_bill'])
sns.histplot(df['total_bill'])
# sns.displot(df['total_bill'],bins = 50 )



```

```python
sns.displot(df['total_bill'],kde = True)
# sns.histplot(df['total_bill'],kde = True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'])
```

# Jointplot

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'scatter' )
```
