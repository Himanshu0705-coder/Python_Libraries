# Plotly and Cufflinks
- Plotly is a python library used to create interactive graphs.

- Cufflinks connects plotly with pandas to create graphs and charts of dataframes directly.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from plotly.offline import iplot
import cufflinks as cf
import plotly.express as px
```

```python
df = sns.load_dataset('tips')
df
```

```python


fig=px.line(df, y="tip")
fig.show()
```

```python

fig = px.bar(df, y="tip")
fig.show()
```

```python
fig = px.scatter(df,y='total_bill')
fig.show()
```

```python
fig = px.histogram(df,y=['sex'])
fig.show()
```

```python
fig = px.box(df,y=['tip'])
fig.show()
```
