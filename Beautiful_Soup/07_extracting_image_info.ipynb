# Extracting Image information in Web Page using Python

```python
import requests
from bs4 import BeautifulSoup
```

```python
web = requests.get("https://www.naukri.com/")
```

```python
web.content
```

```python
soup = BeautifulSoup(web.content , "html.parser")
soup
```

```python
print(soup.prettify())
```

```python
img = soup.find_all("img",src ="//static.naukimg.com/s/0/0/i/naukri-identity/naukriLogo.svg")
img
```

```python
for i in img:
    print(i.get("src"))
```
